```python
import pandas as pd
from generador_pdf import FacturaIELC

# 1. EJECUTAR EL SCRIPT DE CAPTURA (Esto crea la variable factura_final)
%run "creativite absoulute.py"

# 2. ORGANIZAR LOS DATOS CAPTURADOS
df_ventas = pd.DataFrame({
    'Producto': almacen_producto,
    'Precio': precio_unit,
    'Cantidad': almacen_cantidad,
    'Subtotal': lista_subtotales
})

subtotal_p = sum(lista_subtotales)
total_frais = sum(frais_tva) + sum(manut_depot) + sum(march_div)

# 3. DEFINIR LA FUNCIÓN MULTILINGÜE
def generar_documento_final(nombre_archivo, empresa="IELC", idioma="ES", incluir_frais=False):
    pdf = FacturaIELC(empresa=empresa, idioma=idioma)
    pdf.add_page()
    
    labels = {
        "ES": {"fac": "FACTURA NÚMERO", "fec": "FECHA", "cli": "CLIENTE", "tot": "TOTAL ", "sig_c": "firma cliente", "sig_a": "Firma Autorizada"},
        "FR": {"fac": "FACTURE NUMÉRO", "fec": "DATE", "cli": "CLIENT", "tot": "TOTAL", "sig_c": " signature(Client)", "sig_a": "Signature Autorisée"}
    }
    l = labels[idioma]

    pdf.set_font("helvetica", "B", 11)
    pdf.cell(0, 7, f"{l['fac']}: {factura_final}", ln=1)
    pdf.cell(0, 7, f"{l['fec']}: {fecha_limpia}", ln=1)
    pdf.cell(0, 7, f"{l['cli']}: {nombre_apellido1} {nombre_apellido2}", ln=1)
    pdf.ln(5)
    
    pdf.tabla_productos(df_ventas)
    
    total_final = subtotal_p + (total_frais if incluir_frais else 0)
    
    pdf.ln(10)
    pdf.set_fill_color(230, 240, 255)
    pdf.cell(140, 10, l['tot'], 0, 0, "R", True)
    pdf.cell(40, 10, f"{total_final:,.2f}", 1, 1, "R", True)
    
    pdf.ln(20)
    y = pdf.get_y()
    pdf.line(20, y, 70, y); pdf.set_xy(20, y+2); pdf.cell(50, 5, l['sig_c'], 0, 0, "C")
    pdf.line(130, y, 180, y); pdf.set_xy(130, y+2); pdf.cell(50, 5, l['sig_a'], 0, 0, "C")
    
    pdf.output(nombre_archivo)

# 4. GENERAR LOS ARCHIVOS
# Generar versión Creativite Absolute en Francés
generar_documento_final(f"Facture_CA_{factura_final}.pdf", empresa="CREATIVITE", idioma="FR", incluir_frais=True)

# Opcional: Generar versión IELC en Español
generar_documento_final(f"Factura_IELC_{factura_final}.pdf", empresa="IELC", idioma="ES", incluir_frais=False)

print("🚀 ¡Éxito! Las facturas han sido generadas.")
```
